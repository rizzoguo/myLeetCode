* 17 Letter Combinations of a Phone Number

```python
* Time complexity: O(4^n * n)
* Space complexity: O(n)
class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        res = []
        digitToChar = {
            '2': 'abc',
            '3': 'def',
            '4': 'ghi',
            '5': 'jkl',
            '6': 'mno',
            '7': 'pqrs',
            '8': 'tuv',
            '9': 'wxyz',
        }

        def backtracking(index, curChar):
            if len(curChar) == len(digits):
                res.append(curChar)
                return
            
            for c in digitToChar[digits[index]]:
                backtracking(index + 1, curChar + c)
        
        if digits:
            backtracking(0, '')
        return res
```
