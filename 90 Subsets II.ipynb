* 90 Subsets II

```python
# time complexity: O(n⋅2^n)
# space complexity: O(n⋅2^n)
class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        res = []
        path = []

        def dfs(start):
            res.append(path.copy())

            for i in range(start, len(nums)):
                if i > start and nums[i] == nums[i - 1]:
                    continue
                
                path.append(nums[i])
                dfs(i + 1)
                path.pop()
        
        dfs(0)
        return res
```
