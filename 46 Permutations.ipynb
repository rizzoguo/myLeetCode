* 46 Permutations

```python
* time complexity: O(n⋅n!)
* space complexity: O(n⋅n!)
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        res = []
        perm = []
        used = [False] * len(nums)

        def dfs():
            if len(perm) == len(nums):
                res.append(perm.copy())
                return 
            
            for i in range(len(nums)):
                if not used[i]:
                    perm.append(nums[i])
                    used[i] = True
                    dfs()
                    perm.pop()
                    used[i] = False
        
        dfs()
        return res
```
