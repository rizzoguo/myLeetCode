* 1820 Maximum Number of Accepted Invitations

```python
*time complexity: O(m*n*n)
* space complexity: O(m*n)
class Solution:
    def maximumInvitations(self, grid: List[List[int]]) -> int:
        m, n = len(grid), len(grid[0])

        match = {}

        def dfs(boy, visited):
            for girl in range(n):
                if grid[boy][girl] == 1 and girl not in visited:
                    visited.add(girl)

                    if girl not in match or dfs(match[girl], visited):
                        match[girl] = boy
                        return True
            return False
        
        res = 0

        for boy in range(m):
            visited = set()

            if dfs(boy, visited):
                res += 1
        
        return res
```
