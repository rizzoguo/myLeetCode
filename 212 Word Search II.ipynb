```python
*time complexity:O(rows⋅cols⋅4^maxLen)
*space complexity:O(W+rows⋅cols)

class TrieNode:
    def __init__(self):
        self.children = {}
        self.isWord = False

class Solution:
    def findWords(self, board: List[List[str]], words: List[str]) -> List[str]:
        root = TrieNode()
        for word in words:
            cur = root
            for c in word:
                if c not in cur.children:
                    cur.children[c] = TrieNode()
                cur = cur.children[c]
            cur.isWord = True
        
        rows, cols = len(board), len(board[0])
        res, visit = set(), set()

        def dfs(r, c, node, curWord):
            if (r < 0 or c < 0 or r >= rows or c >= cols or 
                (r, c) in visit or board[r][c] not in node.children):
                return
            
            visit.add((r, c))
            curWord += board[r][c]
            node = node.children[board[r][c]]
            if node.isWord:
                res.add(curWord)
            
            dfs(r + 1, c, node, curWord)
            dfs(r, c + 1, node, curWord)
            dfs(r - 1, c, node, curWord)
            dfs(r, c - 1, node, curWord)
            visit.remove((r, c))

        for r in range(rows):
            for c in range(cols):
                dfs(r, c, root, '')
        
        return list(res)
```
