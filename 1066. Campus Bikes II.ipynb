* 1066. Campus Bikes II

```python
*time complexity: O(n * 2^m)
*space complexity: O(n*2^m)
class Solution:
    def assignBikes(self, workers: List[List[int]], bikes: List[List[int]]) -> int:
        memo = {}

        def dfs(workerIndex, mask):
            if workerIndex == len(workers):
                return 0
            
            if (workerIndex, mask) in memo:
                return memo[(workerIndex, mask)]
            
            ans = float('inf')

            for bikeIndex in range(len(bikes)):
                if not (mask & (1 << bikeIndex)):
                    wx, wy = workers[workerIndex]
                    bx, by = bikes[bikeIndex]

                    dist = abs(wx - bx) + abs(wy - by)

                    ans = min(ans, dist + dfs(workerIndex + 1, 
                                            mask | (1 << bikeIndex)))
            
            memo[(workerIndex, mask)] = ans
            return ans
        
        return dfs(0, 0)
```
