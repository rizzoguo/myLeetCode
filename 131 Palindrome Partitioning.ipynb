* 131 Palindrome Partitioning

```python
time complexity: n * 2^n
space complexity: n
class Solution:
    def partition(self, s: str) -> List[List[str]]:
        res, part = [], []
        def dfs(j, i):
            if i >= len(s):
                if i == j:
                    res.append(part.copy())
                return
            
            if self.isPali(s, j, i):
                part.append(s[j : i + 1])
                dfs(i + 1, i + 1)
                part.pop()
            dfs(j, i + 1)
        
        dfs(0, 0)
        return res

    
    def isPali(self, s, l, r):
        while l < r:
            if s[l] != s[r]:
                return False
            l, r = l + 1, r - 1
        return True
```
