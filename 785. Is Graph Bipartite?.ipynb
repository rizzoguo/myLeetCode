* 785. Is Graph Bipartite?

```python
*time complexity: O(v + e)
*space complexity: O(v)
class Solution:
    def isBipartite(self, graph: List[List[int]]) -> bool:
        color = {}

        def dfs(node, c):
            if node in color:
                return color[node] == c
            
            color[node] = c

            for nei in graph[node]:
                if not dfs(nei, -c):
                    return False
            
            return True
        
        for node in range(len(graph)):
            if node not in color:
                if not dfs(node, -1):
                    return False
        
        return True
```
