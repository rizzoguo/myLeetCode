* 886. Possible Bipartition

```python
*time complexity: O(n + e)
* space complexity: O(n + e)
class Solution:
    def possibleBipartition(self, n: int, dislikes: List[List[int]]) -> bool:
        graph = defaultdict(list)

        for a, b in dislikes:
            graph[a].append(b)
            graph[b].append(a)
        
        color = {}

        def dfs(node, c):
            if node in color:
                return color[node] == c
            
            color[node] = c

            for nei in graph[node]:
                if not dfs(nei, -c):
                    return False
            return True
        
        for person in range(n):
            if person not in color:
                if not dfs(person, 1):
                    return False
        
        return True
```
