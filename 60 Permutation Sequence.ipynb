* 60 Permutation Sequence

```python
*time complexity: O(n^2)
* space complexity: O(n)
def getPermutation(self, n: int, k: int) -> str:
    nums = [str(i) for i in range(1, n + 1)]
    k -= 1
    res = ''

    while nums:
        fact = math.factorial(len(nums - 1))
        index = k // fact
        res += nums.pop(index)
        k %= fact
    
    return res
```
